# Case Intro
Term deposits are a major source of income for a bank. A term deposit is a cash investment held at a financial institution. Your money is invested for an agreed rate of interest over a fixed amount of time, or term. The bank has various outreach plans to sell term deposits to their customers such as email marketing, advertisements, telephonic marketing, and digital marketing.

Telephonic marketing campaigns still remain one of the most effective way to reach out to people. However, they require huge investment as large call centers are hired to actually execute these campaigns. Hence, it is crucial to identify the customers most likely to convert beforehand so that they can be specifically targeted via call.

The data is related to direct marketing campaigns (phone calls) of a Portuguese banking institution. The classification goal is to predict if the client will subscribe to a term deposit (variable y).

Content
The data is related to the direct marketing campaigns of a Portuguese banking institution. The marketing campaigns were based on phone calls. Often, more than one contact to the same client was required, in order to access if the product (bank term deposit) would be ('yes') or not ('no') subscribed by the customer or not. The data folder contains two datasets:-

Bank.csv: 45,211 rows and 18 columns ordered by date (from May 2008 to November 2010)

Detailed Column Descriptions
bank client data:

1 - age (numeric)

2 - job : type of job (categorical: "admin.","unknown","unemployed","management","housemaid","entrepreneur","student",
"blue-collar","self-employed","retired","technician","services")

3 - marital : marital status (categorical: "married","divorced","single"; note: "divorced" means divorced or widowed)

4 - education (categorical: "unknown","secondary","primary","tertiary")

5 - default: has credit in default? (binary: "yes","no")

6 - balance: average yearly balance, in euros (numeric)

7 - housing: has housing loan? (binary: "yes","no")

8 - loan: has personal loan? (binary: "yes","no")
# related with the last contact of the current campaign:
9 - contact: contact communication type (categorical: "unknown","telephone","cellular")
10 - day: last contact day of the month (numeric)

11 - month: last contact month of year (categorical: "jan", "feb", "mar", …, "nov", "dec")

12 - duration: last contact duration, in seconds (numeric)

# other attributes:
13 - campaign: number of contacts performed during this campaign and for this client (numeric, includes last contact)

14 - pdays: number of days that passed by after the client was last contacted from a previous campaign (numeric, -1 means client was not previously contacted)

15 - previous: number of contacts performed before this campaign and for this client (numeric)

16 - poutcome: outcome of the previous marketing campaign (categorical: "unknown","other","failure","success")

Output variable (desired target):

17 - y - has the client subscribed a term deposit? (binary: "yes","no")

Missing Attribute Values: None


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('https://raw.githubusercontent.com/ogut77/DataScience/main/data/Bank.csv',sep = ';')
df
```

```python
#For object check the data
for cn in df.columns:
  if(df[cn].dtype==object):
    print(df[cn].value_counts())
```

```python
# Following function converts non-numeric variables (e.g., 'category', 'object') into numeric using label encoding
# Note:
#Label encoding converts categorical values into integer codes. Each unique category is assigned a distinct number, such as 1, 2, 3, 4, etc.
#Example : In our data, education variable has following values:{ 'primary' 'secondary', 'tertiary', 'unknown'}
#and it is mapped { 'primary': 0, 'secondary': 1, 'tertiary': 2, 'unknown': 3}

#  education  education_encoded
#  secondary                  1
#   tertiary                  2
#    primary                  0
#    unknown                  3
#   tertiary                  2
#    primary                  0
#
#Label encoding is generally suitable for tree-based models (e.g., decision trees, random forests, boosting methods).
#However, it may not be appropriate for models where the objective function relies on distance-based calculations, such as neural networks, support vector machines (SVM), or linear regression.
#For nominal features (categories with no intrinsic ordering, e.g., "red," "blue," "green"), label encoding can mislead the model by implying an ordinal relationship where none exists.
#In such cases, one-hot encoding is usually preferred.
#One-hot encoding creates binary columns (dummy variables) for each category in a categorical feature, avoiding the introduction of unintended ordinality.

def Encoder(df):
          from sklearn import preprocessing
          columnsToEncode = list(df.select_dtypes(include=['category','object']))
          le = preprocessing.LabelEncoder()
          for feature in columnsToEncode:
              try:
                  df[feature] = le.fit_transform(df[feature])
              except:
                  print('Error encoding '+feature)
          return df
```

```python
df=Encoder(df)
df
```

```python
y = df['y'] #Output
X = df.drop('y',axis=1)
X
```

```python
#Q1
import pandas as pd
import numpy as np
from sklearn import preprocessing
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

# Separate the target variable (y) from the features (X)
y = df['y']
X = df.drop('y', axis=1)

# Split the data into Training and Testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=17)

models = {
    'Random Forest': RandomForestClassifier(random_state=17),
    'XGBoost': XGBClassifier(random_state=17, eval_metric='logloss'),
    'Light GBM': LGBMClassifier(random_state=17, verbose=-1),
    'Gradient Boosting Classifier': GradientBoostingClassifier(random_state=17)
}

q1_accuracies = {}

# Training and evaluation of each model
print("Test Accuracy with Default Parameters")
for name, model in models.items():
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    acc = accuracy_score(y_test, preds) # accuracy
    q1_accuracies[name] = acc
    print(f"{name}: {acc:.4f}")

# Model with best accuracy
best_q1_model = max(q1_accuracies, key=q1_accuracies.get)
print(f"\n=> The method with the best default accuracy is {best_q1_model} ({q1_accuracies[best_q1_model]:.4f})")
```

```python
#Q2
!pip install optuna
import optuna
from sklearn.model_selection import cross_val_score

# Suppress optuna trial logs to keep the output clean
optuna.logging.set_verbosity(optuna.logging.WARNING)

def objective(trial):
    # Suggestion of which classifier to use for this trial
    classifier_name = trial.suggest_categorical("classifier", ["Random Forest", "XGBoost", "Light GBM", "Gradient Boosting Classifier"])

    # Total number of columns
    n_cols = X_train.shape[1]

    # Now we suggest hyperparameters based on the chosen classifier
    if classifier_name == "Random Forest":
        max_depth = trial.suggest_int("max_depth", 2, n_cols)
        max_features = trial.suggest_int("max_features", 2, n_cols)
        model = RandomForestClassifier(max_depth=max_depth, max_features=max_features, random_state=17)

    elif classifier_name == "XGBoost":
        max_depth = trial.suggest_int("max_depth", 2, n_cols)
        learning_rate = trial.suggest_float("learning_rate", 0.001, 0.3, log=True)
        model = XGBClassifier(max_depth=max_depth, learning_rate=learning_rate, random_state=17, eval_metric='logloss')

    elif classifier_name == "Light GBM":
        max_depth = trial.suggest_int("max_depth", 2, n_cols)
        learning_rate = trial.suggest_float("learning_rate", 0.001, 0.3, log=True)
        model = LGBMClassifier(max_depth=max_depth, learning_rate=learning_rate, random_state=17, verbose=-1)

    else:
        max_depth = trial.suggest_int("max_depth", 2, n_cols)
        learning_rate = trial.suggest_float("learning_rate", 0.001, 0.3, log=True)
        model = GradientBoostingClassifier(max_depth=max_depth, learning_rate=learning_rate, random_state=17)

    #  Cross-Validation (CV=3)
    # n_jobs=-1 uses all CPU cores to speed up the cross-validation
    score = cross_val_score(model, X_train, y_train, cv=3, scoring="accuracy", n_jobs=-1)

    # Mean accuracy across the 3 folds
    return score.mean()

# Cross-validation accuracy and we run here 100 trials
study = optuna.create_study(direction="maximize")
study.optimize(objective, n_trials=100)

print("Best Method and Parameters (CV=3)")
best_params = study.best_params
best_classifier = best_params.pop('classifier') # Extract the best classifier name

print(f"Best Classifier: {best_classifier}")
print(f"Best Parameters: {best_params}")
print(f"Highest CV Accuracy: {study.best_value:.4f}")

# Evaluate Best Model on Test Data

# Winning model with the discovered best parameters
if best_classifier == "Random Forest":
    best_model = RandomForestClassifier(max_depth=best_params['max_depth'], max_features=best_params['max_features'], random_state=17)
elif best_classifier == "XGBoost":
    best_model = XGBClassifier(max_depth=best_params['max_depth'], learning_rate=best_params['learning_rate'], random_state=17, eval_metric='logloss')
elif best_classifier == "Light GBM":
    best_model = LGBMClassifier(max_depth=best_params['max_depth'], learning_rate=best_params['learning_rate'], random_state=17, verbose=-1)
else:
    best_model = GradientBoostingClassifier(max_depth=best_params['max_depth'], learning_rate=best_params['learning_rate'], random_state=17)

# Full training set
best_model.fit(X_train, y_train)

# Prediction on test set
best_preds = best_model.predict(X_test)
best_test_accuracy = accuracy_score(y_test, best_preds)

print("\n--- Final Evaluation on Test Data ---")
print(f"Optimized {best_classifier} Test Accuracy: {best_test_accuracy:.4f}")

# Compare with the same model's default parameter accuracy from Q1
default_accuracy = q1_accuracies[best_classifier]
improvement = best_test_accuracy - default_accuracy

print(f"Default {best_classifier} Test Accuracy (from Q1): {default_accuracy:.4f}")
print(f"Improvement: {improvement:.4f}")
if improvement > 0:
    print("Conclusion: Yes, hyperparameter optimization improved the model's performance on unseen test data.")
else:
    print("Conclusion: The default parameters actually generalized better (or tied) on the test data compared to the CV-optimized parameters.")
```

Q3)Using Linear Regression,Decison Tree Random Forest,XGBoost, Light GBM and Gradient Boosting Classifier with default parameters (no parameter specifications except random_state) calculate R2 statistics on test data. Which method gives the best accuracy on test data

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import r2_score
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor

models = {
    'Linear Regression': LinearRegression(),
    'Decision Tree': DecisionTreeRegressor(random_state=17),
    'Random Forest': RandomForestRegressor(random_state=17),
    'XGBoost': XGBRegressor(random_state=17),
    'Light GBM': LGBMRegressor(random_state=17, verbose=-1),
    'Gradient Boosting': GradientBoostingRegressor(random_state=17)
}

q3_r2_scores = {}

print("--- R2 Scores with Default Parameters ---")
for name, model in models.items():
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    r2 = r2_score(y_test, preds)
    q3_r2_scores[name] = r2
    print(f"{name}: R2 = {r2:.4f}")

best_q3_model = max(q3_r2_scores, key=q3_r2_scores.get)
print(f"\n=> The method with the best default R2 score is {best_q3_model} ({q3_r2_scores[best_q3_model]:.4f})")
```

```python
import optuna
from sklearn.model_selection import cross_val_score

optuna.logging.set_verbosity(optuna.logging.WARNING)

def objective(trial):
    regressor_name = trial.suggest_categorical("regressor", ["Random Forest", "XGBoost", "Light GBM", "Gradient Boosting Regressor"])
    n_cols = X_train.shape[1]

    if regressor_name == "Random Forest":
        max_depth = trial.suggest_int("max_depth", 2, n_cols)
        max_features = trial.suggest_int("max_features", 2, n_cols)
        model = RandomForestRegressor(max_depth=max_depth, max_features=max_features, random_state=17)

    elif regressor_name == "XGBoost":
        max_depth = trial.suggest_int("max_depth", 2, n_cols)
        learning_rate = trial.suggest_float("learning_rate", 0.001, 0.3, log=True)
        model = XGBRegressor(max_depth=max_depth, learning_rate=learning_rate, random_state=17)

    elif regressor_name == "Light GBM":
        max_depth = trial.suggest_int("max_depth", 2, n_cols)
        learning_rate = trial.suggest_float("learning_rate", 0.001, 0.3, log=True)
        model = LGBMRegressor(max_depth=max_depth, learning_rate=learning_rate, random_state=17, verbose=-1)

    else: # Gradient Boosting Regressor
        max_depth = trial.suggest_int("max_depth", 2, n_cols)
        learning_rate = trial.suggest_float("learning_rate", 0.001, 0.3, log=True)
        model = GradientBoostingRegressor(max_depth=max_depth, learning_rate=learning_rate, random_state=17)

    score = cross_val_score(model, X_train, y_train, cv=3, scoring="r2", n_jobs=-1)
    return score.mean()

study = optuna.create_study(direction="maximize")
study.optimize(objective, n_trials=100)

print("--- Best Method & Parameters (CV=3) ---")
best_params = study.best_params
best_regressor = best_params.pop('regressor')

print(f"Best Regressor: {best_regressor}")
print(f"Best Parameters: {best_params}")
print(f"Highest CV R2: {study.best_value:.4f}")

if best_regressor == "Random Forest":
    best_model = RandomForestRegressor(max_depth=best_params['max_depth'], max_features=best_params['max_features'], random_state=17)
elif best_regressor == "XGBoost":
    best_model = XGBRegressor(max_depth=best_params['max_depth'], learning_rate=best_params['learning_rate'], random_state=17)
elif best_regressor == "Light GBM":
    best_model = LGBMRegressor(max_depth=best_params['max_depth'], learning_rate=best_params['learning_rate'], random_state=17, verbose=-1)
else:
    best_model = GradientBoostingRegressor(max_depth=best_params['max_depth'], learning_rate=best_params['learning_rate'], random_state=17)

best_model.fit(X_train, y_train)

best_preds = best_model.predict(X_test)
best_test_r2 = r2_score(y_test, best_preds)

print("\n---  Final Evaluation on Test Data ---")
print(f"Optimized {best_regressor} Test R2: {best_test_r2:.4f}")

q3_name_map = {"Gradient Boosting Regressor": "Gradient Boosting"}
default_name = q3_name_map.get(best_regressor, best_regressor)

default_r2 = q3_r2_scores[default_name]
improvement = best_test_r2 - default_r2

print(f"Default {default_name} Test R2 (from Q3): {default_r2:.4f}")
print(f"Improvement over default: {improvement:.4f}")
```
